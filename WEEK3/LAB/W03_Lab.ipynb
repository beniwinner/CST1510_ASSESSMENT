### D6. A default value

1. Define `status_of(percent, warning_at=90)` that returns:
   - `"OVER LIMIT"` if `percent` is 100 or more
   - `"WARNING"` if `percent` is `warning_at` or more
   - `"OK"` otherwise
2. Print `status_of(95)`, then `status_of(85)`, then `status_of(85, 80)`.

**Expected output:** `WARNING`, `OK`, `WARNING`

````python

def status_of(percent, warning_at=90):
    """Return the status based on the percentage."""
    if percent >= 100:
        return "OVER LIMIT"
    elif percent >= warning_at:
        return "WARNING"
    else:
        return "OK"
print(status_of(95))
print(status_of(85))
print(status_of(85, 80))
````

### D8. `*args` and `**kwargs`

a) Define `total(*numbers)` that returns the sum of all the numbers passed in. Print `total(1, 2, 3)`,
   `total(10, 20, 30, 40)` and `total()`.

b) Define `configure(**settings)` that loops over `settings.items()` and prints each one as `key = value`.
   Call it with `configure(host="10.0.0.1", port=8080)`.

**Expected output:**
```
6
100
0

```

````python
# D8
def total(*numbers):
    """Return the sum of all numbers provided."""
    return sum(numbers)
print(total(1, 2, 3))
print(total(10, 20, 30, 40))
print(total())
def configure(**settings):
    """Print each setting and its value."""
    for key, value in settings.items():
        print(key, "=", value)
configure(host="10.0.0.1", port=8080)
````

### D12. Returning more than one value

1. Define `check_record(value, limit)` that works out `difference = value - limit` and
   `percent = (value / limit) * 100`, and returns **both**: `return difference, percent`
2. Call it with `check_record(87.0, 100.0)` and store the results in two variables.
3. Print them.

**Expected output:**
```
Difference: -13.0
Percent: 87.0
```

````python

def check_record(value, limit):
    """Return the difference and percentage for a record."""
    difference = value - limit
    percent = (value / limit) * 100
    return difference, percent

difference, percent = check_record(87.0, 100.0)

print("Difference:", difference)
print("Percent:", percent)
````

### C1. Wire two functions together

Using `percent_of` from D3 and `status_of` from D6 **together**, print a status for each of these
three pairs. Do not write the percentage or status logic a second time.

| value | limit | Expected output |
|---|---|---|
| 87 | 100 | `OK` |
| 120 | 100 | `OVER LIMIT` |
| 45 | 100 | `OK` |

````python


def percent_of(part, total):
    """Return part as a percentage of total."""
    return (part / total) * 100


def status_of(percent, warning_at=90):
    """Return the status based on the percentage."""
    if percent >= 100:
        return "OVER LIMIT"
    elif percent >= warning_at:
        return "WARNING"
    else:
        return "OK"


records = [(87, 100), (120, 100), (45, 100)]

for value, limit in records:
    percent = percent_of(value, limit)
    print(status_of(percent))
````

### C2. How many are over?

Write `count_over(*values, limit=100)` with a one-line docstring. It returns how many of the values
are over the limit.

| Call | Expected output |
|---|---|
| `count_over(87, 120, 45, 101)` | `2` |
| `count_over(87, 120, 45, 101, limit=50)` | `3` |
| `count_over()` | `0` |

````python
# C2
def count_over(*values, limit=100):
    """Return the number of values that exceed the limit."""
    count = 0

    for value in values:
        if value > limit:
            count += 1

    return count
print(count_over(87, 120, 45, 101))
print(count_over(87, 120, 45, 101, limit=50))
print(count_over())
````
